# MAE / MFE Trade Click Notebook

This notebook performs the first requested task for the MA Cross week sample:

- load MT5 trade history from `ReportHistory-week.xlsx`
- load XAUUSD bid/ask ticks from `XAUUSD_week.csv`
- exclude the carryover trade opened before the tick file starts
- calculate MAE and MFE from the full tick path of each trade
- render a clickable MAE/MFE scatter plot; clicking a point redraws the related trade chart

The retest with statistically-derived SL/TP settings is intentionally left for the next notebook once the correct backtest code/config path is provided.


```python
from pathlib import Path
import json
import math
import html
import warnings

import numpy as np
import pandas as pd
try:
    from IPython.display import HTML, display
except ImportError:
    class HTML(str):
        pass

    def display(value):
        if isinstance(value, HTML):
            print("<HTML output omitted outside Jupyter>")
        else:
            print(value)

warnings.filterwarnings("ignore", category=UserWarning, module="openpyxl")

NOTEBOOK_DIR = Path.cwd()
TICK_FILE = NOTEBOOK_DIR / "XAUUSD_week.csv"
REPORT_FILE = NOTEBOOK_DIR / "ReportHistory-week.xlsx"
if not TICK_FILE.exists():
    TICK_FILE = Path(r"C:\Users\HP\Downloads\XAUUSD_week.csv")
if not REPORT_FILE.exists():
    REPORT_FILE = Path(r"C:\Users\HP\Downloads\ReportHistory-week.xlsx")

SYMBOL = "XAUUSD"
POINT_SIZE = 0.01
MAX_CHART_POINTS_PER_TRADE = 500

assert TICK_FILE.exists(), f"Missing tick file: {TICK_FILE}"
assert REPORT_FILE.exists(), f"Missing report file: {REPORT_FILE}"
```

```python
def find_section(raw: pd.DataFrame, label: str) -> int:
    matches = raw.apply(lambda row: row.astype(str).str.strip().eq(label).any(), axis=1)
    idx = raw.index[matches].tolist()
    if not idx:
        raise ValueError(f"Could not find section {label!r}")
    return int(idx[0])


def load_mt5_positions(path: Path, symbol: str) -> pd.DataFrame:
    raw = pd.read_excel(path, sheet_name="Sheet1", header=None)
    positions_row = find_section(raw, "Positions")
    orders_row = find_section(raw, "Orders")

    rows = raw.iloc[positions_row + 2 : orders_row].copy()
    rows = rows[rows[0].notna()].copy()
    rows = rows.iloc[:, :13]
    rows.columns = [
        "entry_time",
        "position",
        "symbol",
        "type",
        "volume",
        "entry_price",
        "sl",
        "tp",
        "exit_time",
        "exit_price",
        "commission",
        "swap",
        "profit",
    ]

    rows["symbol"] = rows["symbol"].astype(str).str.strip()
    rows["type"] = rows["type"].astype(str).str.lower().str.strip()
    rows = rows[rows["symbol"].eq(symbol)].copy()

    rows["entry_time"] = pd.to_datetime(rows["entry_time"], format="%Y.%m.%d %H:%M:%S", errors="coerce")
    rows["exit_time"] = pd.to_datetime(rows["exit_time"], format="%Y.%m.%d %H:%M:%S", errors="coerce")
    for col in ["volume", "entry_price", "sl", "tp", "exit_price", "commission", "swap", "profit"]:
        rows[col] = pd.to_numeric(rows[col], errors="coerce")

    rows = rows.dropna(subset=["entry_time", "exit_time", "entry_price", "exit_price", "profit"])
    rows = rows.sort_values(["entry_time", "position"]).reset_index(drop=True)
    rows["original_trade_id"] = rows.index
    return rows


def load_ticks(path: Path) -> pd.DataFrame:
    if path.stat().st_size < 50_000_000:
        raise RuntimeError(
            f"Tick file looks incomplete: {path} is only {path.stat().st_size:,} bytes. "
            "Use the full XAUUSD_week.csv export, expected around 86 MB."
        )

    ticks = pd.read_csv(path, sep="\t")
    ticks.columns = [str(c).strip().lower().replace("<", "").replace(">", "") for c in ticks.columns]
    required = {"date", "time", "bid", "ask"}
    missing = required.difference(ticks.columns)
    if missing:
        raise ValueError(f"Tick file missing columns: {sorted(missing)}")

    ticks["datetime"] = pd.to_datetime(
        ticks["date"].astype(str) + " " + ticks["time"].astype(str),
        format="%Y.%m.%d %H:%M:%S.%f",
        errors="coerce",
    )
    ticks["bid"] = pd.to_numeric(ticks["bid"], errors="coerce")
    ticks["ask"] = pd.to_numeric(ticks["ask"], errors="coerce")
    ticks = ticks.dropna(subset=["datetime", "bid", "ask"]).sort_values("datetime").reset_index(drop=True)
    ticks = ticks[(ticks["bid"] > 1000) & (ticks["ask"] > 1000) & (ticks["ask"] >= ticks["bid"])].copy()
    if ticks.empty:
        raise RuntimeError("No valid XAUUSD bid/ask ticks were found after cleaning.")
    ticks["mid"] = (ticks["bid"] + ticks["ask"]) / 2.0
    return ticks[["datetime", "bid", "ask", "mid"]]
```

```python
positions_raw = load_mt5_positions(REPORT_FILE, SYMBOL)
ticks = load_ticks(TICK_FILE)

tick_start = ticks["datetime"].min()
tick_end = ticks["datetime"].max()

excluded = positions_raw[positions_raw["entry_time"] < tick_start].copy()
positions = positions_raw[
    (positions_raw["entry_time"] >= tick_start)
    & (positions_raw["exit_time"] <= tick_end)
].copy().reset_index(drop=True)
positions["trade_id"] = positions.index

print(f"Tick coverage: {tick_start} to {tick_end}")
print(f"Raw closed positions: {len(positions_raw):,}")
print(f"Excluded carryover positions opened before tick coverage: {len(excluded):,}")
display(excluded[["original_trade_id", "entry_time", "exit_time", "type", "entry_price", "exit_price", "sl", "tp", "profit"]])
print(f"Included positions for MAE/MFE: {len(positions):,}")
display(positions.head())
```

```python
def downsample_indices(length: int, max_points: int) -> np.ndarray:
    if length <= max_points:
        return np.arange(length)
    return np.unique(np.linspace(0, length - 1, max_points).round().astype(int))


def calculate_trade_excursions(trades: pd.DataFrame, ticks: pd.DataFrame) -> pd.DataFrame:
    tick_times = ticks["datetime"].to_numpy()
    bid = ticks["bid"].to_numpy(dtype=float)
    ask = ticks["ask"].to_numpy(dtype=float)
    mid = ticks["mid"].to_numpy(dtype=float)
    records = []

    for row in trades.itertuples(index=False):
        start = np.searchsorted(tick_times, np.datetime64(row.entry_time), side="left")
        # MT5 report timestamps are second-precision while ticks are millisecond-precision.
        # Include ticks inside the reported exit second so stop/TP touches are not missed.
        exit_boundary = row.exit_time + pd.Timedelta(seconds=1)
        end = np.searchsorted(tick_times, np.datetime64(exit_boundary), side="left")
        if end <= start:
            continue

        bid_w = bid[start:end]
        ask_w = ask[start:end]
        mid_w = mid[start:end]
        time_w = tick_times[start:end]

        if row.type == "buy":
            closeable = bid_w
            favorable = bid_w - row.entry_price
            adverse = row.entry_price - bid_w
        elif row.type == "sell":
            closeable = ask_w
            favorable = row.entry_price - ask_w
            adverse = ask_w - row.entry_price
        else:
            continue

        mae_points = max(0.0, float(np.nanmax(adverse)) / POINT_SIZE)
        mfe_points = max(0.0, float(np.nanmax(favorable)) / POINT_SIZE)
        path_points = np.where(row.type == "buy", (bid_w - row.entry_price) / POINT_SIZE, (row.entry_price - ask_w) / POINT_SIZE)

        chart_idx = downsample_indices(len(time_w), MAX_CHART_POINTS_PER_TRADE)

        records.append(
            {
                "trade_id": int(row.trade_id),
                "original_trade_id": int(row.original_trade_id),
                "position": str(row.position),
                "symbol": row.symbol,
                "type": row.type,
                "volume": float(row.volume),
                "entry_time": row.entry_time,
                "exit_time": row.exit_time,
                "duration_min": round((row.exit_time - row.entry_time).total_seconds() / 60.0, 3),
                "entry_price": float(row.entry_price),
                "exit_price": float(row.exit_price),
                "sl": float(row.sl),
                "tp": float(row.tp),
                "profit": float(row.profit),
                "commission": float(row.commission),
                "swap": float(row.swap),
                "mae_points": round(mae_points, 2),
                "mfe_points": round(mfe_points, 2),
                "max_adverse_price": round(float(closeable[int(np.nanargmax(adverse))]), 5),
                "max_favorable_price": round(float(closeable[int(np.nanargmax(favorable))]), 5),
                "min_path_points": round(float(np.nanmin(path_points)), 2),
                "max_path_points": round(float(np.nanmax(path_points)), 2),
                "chart_time": pd.to_datetime(time_w[chart_idx]).strftime("%Y-%m-%d %H:%M:%S.%f").str[:-3].tolist(),
                "chart_bid": np.round(bid_w[chart_idx], 3).tolist(),
                "chart_ask": np.round(ask_w[chart_idx], 3).tolist(),
                "chart_mid": np.round(mid_w[chart_idx], 3).tolist(),
            }
        )

    out = pd.DataFrame(records)
    return out.sort_values("entry_time").reset_index(drop=True)


trades = calculate_trade_excursions(positions, ticks)
print(f"Calculated MAE/MFE for {len(trades):,} trades")
display(trades[["trade_id", "entry_time", "exit_time", "type", "profit", "mae_points", "mfe_points"]].head())
```

```python
summary = {
    "included_trades": len(trades),
    "net_profit": round(float(trades["profit"].sum()), 2),
    "win_rate_pct": round(float((trades["profit"] > 0).mean() * 100), 2),
    "avg_mae_points": round(float(trades["mae_points"].mean()), 2),
    "median_mae_points": round(float(trades["mae_points"].median()), 2),
    "avg_mfe_points": round(float(trades["mfe_points"].mean()), 2),
    "median_mfe_points": round(float(trades["mfe_points"].median()), 2),
}

display(pd.DataFrame([summary]))

quantiles = [0.10, 0.25, 0.50, 0.75, 0.90, 0.95]
mae_mfe_quantiles = trades[["mae_points", "mfe_points"]].quantile(quantiles).rename_axis("quantile")
display(mae_mfe_quantiles)

display(
    trades.groupby(trades["profit"].gt(0).map({True: "winner", False: "loser"}))[["mae_points", "mfe_points", "profit"]]
    .agg(["count", "mean", "median", "min", "max"])
)
```

```python
def build_click_widget(trades: pd.DataFrame) -> HTML:
    payload_cols = [
        "trade_id", "original_trade_id", "position", "type", "volume",
        "entry_time", "exit_time", "duration_min",
        "entry_price", "exit_price", "sl", "tp", "profit",
        "mae_points", "mfe_points", "max_adverse_price", "max_favorable_price",
        "chart_time", "chart_bid", "chart_ask", "chart_mid",
    ]

    payload = trades[payload_cols].copy()
    payload["entry_time"] = payload["entry_time"].dt.strftime("%Y-%m-%d %H:%M:%S")
    payload["exit_time"] = payload["exit_time"].dt.strftime("%Y-%m-%d %H:%M:%S")
    records = payload.to_dict(orient="records")
    data_json = json.dumps(records)

    app_html = f'''
<div id="maeMfeApp" style="font-family: Segoe UI, Arial, sans-serif; color:#1f2933;">
  <div style="display:grid; grid-template-columns: minmax(420px, 1fr) minmax(420px, 1fr); gap:16px; align-items:start;">
    <div>
      <h3 style="margin:0 0 8px;">MAE by Trade</h3>
      <div style="font-size:12px; color:#52606d; margin-bottom:8px;">Maximum adverse excursion. Click a point to render that trade.</div>
      <svg id="maeSvg" width="680" height="380" style="max-width:100%; border:1px solid #d9e2ec; background:white;"></svg>
      <h3 style="margin:18px 0 8px;">MFE by Trade</h3>
      <div style="font-size:12px; color:#52606d; margin-bottom:8px;">Maximum favorable excursion. Click a point to render that trade.</div>
      <svg id="mfeSvg" width="680" height="380" style="max-width:100%; border:1px solid #d9e2ec; background:white;"></svg>
    </div>
    <div>
      <h3 style="margin:0 0 8px;">Selected Trade Chart</h3>
      <div id="tradeMeta" style="font-size:13px; min-height:68px; color:#334e68;"></div>
      <svg id="tradeSvg" width="680" height="600" style="max-width:100%; border:1px solid #d9e2ec; background:white;"></svg>
    </div>
  </div>
</div>
<script>
(function() {{
  const trades = {data_json};
  const maeSvg = document.getElementById("maeSvg");
  const mfeSvg = document.getElementById("mfeSvg");
  const tradeSvg = document.getElementById("tradeSvg");
  const tradeMeta = document.getElementById("tradeMeta");
  const ns = "http://www.w3.org/2000/svg";
  let selectedTradeId = trades.length ? trades[0].trade_id : null;

  function clear(svg) {{ while (svg.firstChild) svg.removeChild(svg.firstChild); }}
  function el(name, attrs, text) {{
    const node = document.createElementNS(ns, name);
    Object.entries(attrs || {{}}).forEach(([k, v]) => node.setAttribute(k, v));
    if (text !== undefined) node.textContent = text;
    return node;
  }}
  function extent(values, padRatio=0.06) {{
    let min = Math.min(...values), max = Math.max(...values);
    if (!Number.isFinite(min) || !Number.isFinite(max)) return [0, 1];
    if (min === max) {{ min -= 1; max += 1; }}
    const pad = (max - min) * padRatio;
    return [Math.max(0, min - pad), max + pad];
  }}
  function scale(value, d0, d1, r0, r1) {{
    return r0 + (value - d0) * (r1 - r0) / (d1 - d0);
  }}
  function drawAxes(svg, margin, xLabel, yLabel, xDomain, yDomain, w, h) {{
    svg.appendChild(el("line", {{x1:margin.l, y1:h-margin.b, x2:w-margin.r, y2:h-margin.b, stroke:"#9fb3c8"}}));
    svg.appendChild(el("line", {{x1:margin.l, y1:margin.t, x2:margin.l, y2:h-margin.b, stroke:"#9fb3c8"}}));
    for (let i=0; i<=5; i++) {{
      const xv = xDomain[0] + (xDomain[1]-xDomain[0])*i/5;
      const x = scale(xv, xDomain[0], xDomain[1], margin.l, w-margin.r);
      svg.appendChild(el("line", {{x1:x, y1:h-margin.b, x2:x, y2:h-margin.b+5, stroke:"#9fb3c8"}}));
      svg.appendChild(el("text", {{x:x, y:h-margin.b+20, "text-anchor":"middle", "font-size":"11", fill:"#52606d"}}, xv.toFixed(0)));
      const yv = yDomain[0] + (yDomain[1]-yDomain[0])*i/5;
      const y = scale(yv, yDomain[0], yDomain[1], h-margin.b, margin.t);
      svg.appendChild(el("line", {{x1:margin.l-5, y1:y, x2:margin.l, y2:y, stroke:"#9fb3c8"}}));
      svg.appendChild(el("text", {{x:margin.l-9, y:y+4, "text-anchor":"end", "font-size":"11", fill:"#52606d"}}, yv.toFixed(0)));
    }}
    svg.appendChild(el("text", {{x:w/2, y:h-12, "text-anchor":"middle", "font-size":"13", fill:"#334e68"}}, xLabel));
    const yText = el("text", {{x:16, y:h/2, "text-anchor":"middle", "font-size":"13", fill:"#334e68", transform:`rotate(-90 16 ${{h/2}})`}}, yLabel);
    svg.appendChild(yText);
  }}
  function renderMetricPlot(svg, metric, title, color) {{
    clear(svg);
    const w = 680;
    const h = 380;
    const margin = {{l:58, r:24, t:22, b:52}};
    const xDomain = [0, Math.max(1, trades.length - 1)];
    const yDomain = extent(trades.map(d => d[metric]));
    drawAxes(svg, margin, "Trade number", title, xDomain, yDomain, w, h);
    const path = trades.map((t, i) => {{
      const x = scale(i, xDomain[0], xDomain[1], margin.l, w-margin.r);
      const y = scale(t[metric], yDomain[0], yDomain[1], h-margin.b, margin.t);
      return `${{i===0 ? "M" : "L"}}${{x.toFixed(2)}},${{y.toFixed(2)}}`;
    }}).join(" ");
    svg.appendChild(el("path", {{d:path, fill:"none", stroke:color, "stroke-width":1.2, opacity:"0.55"}}));
    trades.forEach((t, i) => {{
      const x = scale(i, xDomain[0], xDomain[1], margin.l, w-margin.r);
      const y = scale(t[metric], yDomain[0], yDomain[1], h-margin.b, margin.t);
      const profitable = t.profit > 0;
      const selected = t.trade_id === selectedTradeId;
      const c = el("circle", {{
        cx:x, cy:y, r:selected ? 7 : 4.5,
        fill: selected ? "#102a43" : (profitable ? "#2f9e44" : "#d64545"),
        stroke: selected ? "#102a43" : "white",
        "stroke-width": selected ? 2 : 1,
        style:"cursor:pointer"
      }});
      c.appendChild(el("title", {{}}, `Trade ${{t.trade_id}} | ${{t.type}} | PnL ${{t.profit.toFixed(2)}} | MAE ${{t.mae_points}} | MFE ${{t.mfe_points}}`));
      c.addEventListener("click", () => {{
        selectedTradeId = t.trade_id;
        renderMetricPlot(maeSvg, "mae_points", "MAE (points)", "#d64545");
        renderMetricPlot(mfeSvg, "mfe_points", "MFE (points)", "#2f9e44");
        renderTrade(t);
      }});
      svg.appendChild(c);
    }});
  }}
  function renderTrade(t) {{
    clear(tradeSvg);
    const w = 680, h = 500;
    const margin = {{l:68, r:24, t:24, b:66}};
    const prices = [...t.chart_bid, ...t.chart_ask, t.entry_price, t.exit_price, t.sl, t.tp].filter(Number.isFinite);
    const yDomain = extent(prices, 0.04);
    const n = t.chart_time.length;
    const xDomain = [0, Math.max(1, n-1)];
    drawAxes(tradeSvg, margin, "Trade time path", "Price", xDomain, yDomain, w, h);
    function pathFor(values) {{
      return values.map((v, i) => {{
        const x = scale(i, xDomain[0], xDomain[1], margin.l, w-margin.r);
        const y = scale(v, yDomain[0], yDomain[1], h-margin.b, margin.t);
        return `${{i===0 ? "M" : "L"}}${{x.toFixed(2)}},${{y.toFixed(2)}}`;
      }}).join(" ");
    }}
    function hline(value, color, label, dash) {{
      const y = scale(value, yDomain[0], yDomain[1], h-margin.b, margin.t);
      tradeSvg.appendChild(el("line", {{x1:margin.l, y1:y, x2:w-margin.r, y2:y, stroke:color, "stroke-width":1.5, "stroke-dasharray":dash || ""}}));
      tradeSvg.appendChild(el("text", {{x:w-margin.r-4, y:y-5, "text-anchor":"end", "font-size":"11", fill:color}}, `${{label}} ${{value.toFixed(2)}}`));
    }}
    tradeSvg.appendChild(el("path", {{d:pathFor(t.chart_bid), fill:"none", stroke:"#1d70b8", "stroke-width":1.4}}));
    tradeSvg.appendChild(el("path", {{d:pathFor(t.chart_ask), fill:"none", stroke:"#f59f00", "stroke-width":1.2, opacity:"0.85"}}));
    hline(t.entry_price, "#334e68", "Entry", "4 4");
    hline(t.sl, "#d64545", "SL", "6 4");
    hline(t.tp, "#2f9e44", "TP", "6 4");
    const entryX = margin.l;
    const exitX = w - margin.r;
    tradeSvg.appendChild(el("circle", {{cx:entryX, cy:scale(t.entry_price, yDomain[0], yDomain[1], h-margin.b, margin.t), r:5, fill:"#102a43"}}));
    tradeSvg.appendChild(el("circle", {{cx:exitX, cy:scale(t.exit_price, yDomain[0], yDomain[1], h-margin.b, margin.t), r:5, fill:t.profit > 0 ? "#2f9e44" : "#d64545"}}));
    tradeSvg.appendChild(el("text", {{x:margin.l, y:h-38, "font-size":"11", fill:"#52606d"}}, t.chart_time[0] || ""));
    tradeSvg.appendChild(el("text", {{x:w-margin.r, y:h-38, "text-anchor":"end", "font-size":"11", fill:"#52606d"}}, t.chart_time[n-1] || ""));
    tradeMeta.innerHTML = `
      <strong>Trade ${{t.trade_id}}</strong> | position ${{t.position}} | ${{t.type.toUpperCase()}} | PnL <strong>${{t.profit.toFixed(2)}}</strong><br>
      Entry: ${{t.entry_time}} @ ${{t.entry_price.toFixed(2)}} | Exit: ${{t.exit_time}} @ ${{t.exit_price.toFixed(2)}} | Duration: ${{t.duration_min}} min<br>
      MAE: <strong>${{t.mae_points}}</strong> pts | MFE: <strong>${{t.mfe_points}}</strong> pts
    `;
  }}
  const first = trades[0];
  renderMetricPlot(maeSvg, "mae_points", "MAE (points)", "#d64545");
  renderMetricPlot(mfeSvg, "mfe_points", "MFE (points)", "#2f9e44");
  if (first) renderTrade(first);
}})();
</script>
'''
    return HTML(
        '<iframe '
        'style="width:100%; height:1320px; border:0; border-radius:8px; background:white;" '
        f'srcdoc="{html.escape(app_html, quote=True)}">'
        '</iframe>'
    )


display(build_click_widget(trades))
```
